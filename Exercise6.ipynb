## Sample Submission File

## 1. Train a KNN Regressor

```python
score_list = {}
```

```python
from sklearn.neighbors import KNeighborsRegressor

# Select numerical features and the target variable
numerical_features = df.select_dtypes(include=['int64', 'float64']).columns
X = df[numerical_features].drop('SalePrice', axis=1)
y = df['SalePrice']

# Handle missing values by filling with the mean (if any remain)
X = X.fillna(X.mean())

# Train KNN Regressor
knn = KNeighborsRegressor(n_neighbors=22) # You can tune n_neighbors
knn.fit(X, y)

# Store the score (e.g., R-squared)
score_list['KNN'] = knn.score(X, y)
print(f"KNN Regressor score: {score_list['KNN']}")
```

- Perform cross validation

```python
from sklearn.model_selection import cross_val_score

# Perform cross-validation for KNN Regressor
knn_scores = cross_val_score(knn, X, y, cv=5) # Using 5 folds for cross-validation

print(f"KNN Regressor cross-validation scores: {knn_scores}")
print(f"Average KNN Regressor cross-validation score: {knn_scores.mean()}")
```

## 2. Train a SVM Regression

```python
from sklearn.svm import SVR

# Train SVM Regressor
svr = SVR()
svr.fit(X, y)

# Store the score (e.g., R-squared)
score_list['SVR'] = svr.score(X, y)
print(f"SVM Regressor score: {score_list['SVR']}")
```

- Perform cross validation

```python
from sklearn.model_selection import cross_val_score

# Perform cross-validation for SVR
svr_scores = cross_val_score(svr, X, y, cv=5) # Using 5 folds for cross-validation

print(f"SVR Regressor cross-validation scores: {svr_scores}")
print(f"Average SVR Regressor cross-validation score: {svr_scores.mean()}")
```

## 3. Train a Decision Tree Regression

```python
from sklearn.tree import DecisionTreeRegressor

# Train Decision Tree Regressor
dt = DecisionTreeRegressor(random_state=1)
dt.fit(X, y)

# Store the score (e.g., R-squared)
score_list['Decision Tree'] = dt.score(X, y)
print(f"Decision Tree Regressor score: {score_list['Decision Tree']}")
```

- Perform cross validation

```python
from sklearn.model_selection import cross_val_score

# Perform cross-validation for Decision Tree Regressor
dt_scores = cross_val_score(dt, X, y, cv=5) # Using 5 folds for cross-validation

print(f"Decision Tree Regressor cross-validation scores: {dt_scores}")
print(f"Average Decision Tree Regressor cross-validation score: {dt_scores.mean()}")
```

## 4. Train a Random Forest Regression

```python
from sklearn.ensemble import RandomForestRegressor

# Train Random Forest Regressor
rfr = RandomForestRegressor(n_estimators=50, random_state=1)
rfr.fit(X, y)

# Store the score (e.g., R-squared)
score_list['Random Forest'] = rfr.score(X, y)
print(f"Random Forest Regressor score: {score_list['Random Forest']}")
```

```python
from sklearn.model_selection import cross_val_score

# Perform cross-validation for Random Forest Regressor
rfr_scores = cross_val_score(rfr, X, y, cv=25) # Using 5 folds for cross-validation

print(f"Random Forest Regressor cross-validation scores: {rfr_scores}")
print(f"Average Random Forest Regressor cross-validation score: {rfr_scores.mean()}")
```

## 5. Compare all the performance of all regression models

```python
# put your answer here
print("\nCross-validation Scores:")
print(f"KNN Regressor: {knn_scores.mean()}")
print(f"SVR Regressor: {svr_scores.mean()}")
print(f"Decision Tree Regressor: {dt_scores.mean()}")
print(f"Random Forest Regressor: {rfr_scores.mean()}")
```

## 6. Generate Submission File

Choose the model that has the best performance to generate a submission file.

```python
# Load the sample submission file to get the correct format
sample_submission_url ='https://github.com/robitussin/CCMACLRL_EXERCISES/blob/3fd7d51ffd17863598ac3f44eeefc558171a5b73/dataset/house-prices-advanced-regression-techniques/sample_submission.csv?raw=true'
sf=pd.read_csv(sample_submission_url)

# Load the test data
test_url = 'https://github.com/robitussin/CCMACLRL_EXERCISES/blob/3fd7d51ffd17863598ac3f44eeefc558171a5b73/dataset/house-prices-advanced-regression-techniques/test.csv?raw=true'
dt = pd.read_csv(test_url)

# Apply the same preprocessing steps to the test data as the training data
# Drop the same columns
dt = dt.drop(['Alley', 'PoolQC', 'Fence', 'MiscFeature'], axis=1, errors='ignore')

# Fill missing LotFrontage with the mean (of the test data)
dt['LotFrontage'] = dt['LotFrontage'].fillna(dt['LotFrontage'].mean())

# Convert 'SaleCondition' to numerical in the test data
dt["SaleCondition"] = [1 if each == "Normal" else 0 for each in dt["SaleCondition"]]

# Select numerical features, ensuring they are the same as used for training
numerical_features_train = df.select_dtypes(include=['int64', 'float64']).columns.drop('SalePrice')
X_test = dt[numerical_features_train].copy()

# Fill any remaining missing numerical values in the test features
X_test = X_test.fillna(X_test.mean())


# Use the best performing model (Random Forest Regressor 'rfr') to make predictions
predictions = rfr.predict(X_test)

# Create a submission DataFrame
submission_df = pd.DataFrame({
    'Id': sf['Id'],
    'SalePrice': predictions
})

# Save the submission DataFrame to a CSV file
submission_df.to_csv('submission_file.csv', index=False)
print("Submission file created: submission_file.csv")
```
